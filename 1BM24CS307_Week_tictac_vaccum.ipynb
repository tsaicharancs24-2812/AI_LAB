```python
board = [" " for _ in range(9)]


def display_board():
    print()
    print(board[0], "|", board[1], "|", board[2])
    print("--+---+--")
    print(board[3], "|", board[4], "|", board[5])
    print("--+---+--")
    print(board[6], "|", board[7], "|", board[8])
    print()


def check_winner(player):
    winning_positions = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),
        (0, 4, 8),
        (2, 4, 6)
    ]
    for a, b, c in winning_positions:
        if board[a] == player and board[b] == player and board[c] == player:
            return True

    return False


def computer_move():


    for i in range(9):
        if board[i] == " ":
            board[i] = "O"

            if check_winner("O"):
                return

            board[i] = " "

    for i in range(9):
        if board[i] == " ":
            board[i] = "X"

            if check_winner("X"):
                board[i] = "O"
                return

            board[i] = " "

    for i in range(9):
        if board[i] == " ":
            board[i] = "O"
            return


while True:


    display_board()


    position = int(input("Enter your position (1-9): ")) - 1

    if position < 0 or position > 8:
        print("Invalid position!")
        continue

    if board[position] != " ":
        print("That box is already filled!")
        continue

    board[position] = "X"


    if check_winner("X"):
        display_board()
        print("You win!")
        break


    if " " not in board:
        display_board()
        print("It's a draw!")
        break


    computer_move()


    if check_winner("O"):
        display_board()
        print("Computer wins!")
        break

    if " " not in board:
        display_board()
        print("It's a draw!")
        break
```

```python
import random

room = [random.choice(["Dirty", "Clean"]) for _ in range(2)]
position = 0

while "Dirty" in room:
  print("Rooms: ",room)
  print("Robot Position:", position)

  if(room[position] == "Dirty"):
    print("SUCK")
    room[position] = "Clean"
  elif(position == 0):
    print("ACTION: Move RIGHT")
    position = position + 1
  else:
    print("ACTION: Move Left")
    position = position - 1
print("Rooms:",room, "\nALL ROOMS ARE CLEAN")
```
